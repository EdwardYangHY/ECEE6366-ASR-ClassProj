```python
import os
import shutil
import pandas as pd
```

```python
csv_file = './Data/violin_raw_videos/audio_parts/audio_files_part_1.csv'  # Replace with your CSV file name
df = pd.read_csv(csv_file)
```

```python
destination_dir = './audio_part_split_1'
os.makedirs(destination_dir, exist_ok=True)

# Step 3: Copy each audio file to the new directory
for file_path in df['File Name']:
    # Ensure the file exists before copying
    full_path = os.path.join("./Data/violin_raw_videos", file_path.lstrip('./'))  # Remove './' before joining
    if os.path.isfile(full_path):
        shutil.copy(full_path, destination_dir)
        print(f"Copied: {full_path}")
    else:
        print(f"File not found: {full_path}")

print("All files have been copied to the 'part1' directory.")
```
