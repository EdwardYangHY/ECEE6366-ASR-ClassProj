```python
import json
import os
from pydub import AudioSegment
from tqdm import tqdm
import shutil
import pandas as pd
from glob import glob
```

```python
with open("./Data/violin_raw_videos/violin_annotation.json") as f:
    violin_annotation = json.load(f)
with open("./Data/violin_audios/data_subtitles.json") as f:
    asr_annotation = json.load(f)
```

```python
asr_path = "./Data/violin_audios/data_subtitles.json"
with open(asr_path) as f:
    asr_annotation = json.load(f)
for key in asr_annotation.keys():
    asr_annotation[key]['split']=violin_annotation[key]['split']
with open(asr_path, "w") as f:
    json.dump(asr_annotation, f, indent=4)
```
